```python
from IPython.display import display, IFrame
import os

html = r'''
<!DOCTYPE html>
<html lang="en">
<head>
<meta charset="UTF-8">
<meta name="viewport" content="width=device-width, initial-scale=1.0">

<title>GeoTrack | GPS Trilateration</title>

<style>
* {
    margin: 0;
    padding: 0;
    box-sizing: border-box;
}

html {
    scroll-behavior: smooth;
}

body {
    font-family: Arial, Helvetica, sans-serif;
    background: #07111f;
    color: #eaf2ff;
    line-height: 1.6;
}

/* ================= HEADER ================= */

.header {
    position: sticky;
    top: 0;
    z-index: 1000;

    height: 72px;
    padding: 0 5%;

    display: flex;
    align-items: center;
    gap: 30px;

    background: rgba(7, 17, 31, 0.96);
    border-bottom: 1px solid #1d3148;
    backdrop-filter: blur(10px);
}

.logo {
    display: flex;
    align-items: center;
    gap: 10px;

    font-size: 24px;
    font-weight: bold;
    white-space: nowrap;
}

.logo-icon {
    width: 34px;
    height: 34px;
    border-radius: 50%;

    background: #18d6a5;
    position: relative;
    box-shadow: 0 0 20px rgba(24,214,165,0.4);
}

.logo-icon::before {
    content: "";
    position: absolute;
    width: 12px;
    height: 12px;
    background: #07111f;
    border-radius: 50%;
    left: 11px;
    top: 11px;
}

.accent {
    color: #18d6a5;
}

.team-info {
    display: flex;
    flex-direction: column;
    font-size: 11px;
    color: #8ea4bd;
}

.team-info strong {
    color: #ffffff;
    font-size: 13px;
}

.nav {
    margin-left: auto;
    display: flex;
    gap: 22px;
}

.nav a {
    text-decoration: none;
    color: #9db0c5;
    font-size: 14px;
    transition: 0.3s;
}

.nav a:hover,
.nav a.active {
    color: #18d6a5;
}

.menu-btn {
    display: none;
    background: none;
    border: 0;
    color: white;
    font-size: 25px;
    cursor: pointer;
}

/* ================= GENERAL ================= */

.container {
    width: 90%;
    max-width: 1200px;
    margin: auto;
}

section {
    padding: 90px 0;
}

.section-title {
    text-align: center;
    margin-bottom: 45px;
}

.section-title h2 {
    font-size: 35px;
    margin-bottom: 10px;
}

.section-title p {
    color: #8ea4bd;
}

/* ================= HERO ================= */

.hero {
    min-height: 500px;
    display: flex;
    align-items: center;

    background:
        radial-gradient(circle at 75% 40%, rgba(24,214,165,0.13), transparent 30%),
        radial-gradient(circle at 20% 20%, rgba(55,120,255,0.12), transparent 30%);
}

.hero-content {
    max-width: 750px;
}

.badge {
    display: inline-block;
    padding: 7px 14px;
    border: 1px solid #1f7d6a;
    border-radius: 30px;
    color: #18d6a5;
    font-size: 13px;
    margin-bottom: 20px;
}

.hero h1 {
    font-size: clamp(42px, 7vw, 75px);
    line-height: 1.05;
    margin-bottom: 20px;
}

.hero h1 span {
    color: #18d6a5;
}

.hero p {
    max-width: 700px;
    color: #9db0c5;
    font-size: 18px;
}

.hero-buttons {
    margin-top: 30px;
    display: flex;
    gap: 15px;
}

.btn {
    border: none;
    padding: 13px 22px;
    border-radius: 8px;
    cursor: pointer;
    font-weight: bold;
    text-decoration: none;
    display: inline-block;
}

.btn-primary {
    background: #18d6a5;
    color: #06131d;
}

.btn-secondary {
    border: 1px solid #29435e;
    color: white;
    background: transparent;
}

/* ================= SIMULATOR ================= */

.simulator-grid {
    display: grid;
    grid-template-columns: 330px 1fr;
    gap: 25px;
}

.panel {
    background: #0c1a2b;
    border: 1px solid #1d3148;
    border-radius: 14px;
    padding: 24px;
}

.panel h3 {
    margin-bottom: 20px;
}

.input-group {
    margin-bottom: 18px;
}

.input-group label {
    display: block;
    font-size: 13px;
    color: #9db0c5;
    margin-bottom: 7px;
}

.input-group input {
    width: 100%;
    padding: 11px;
    border-radius: 7px;
    border: 1px solid #29435e;
    background: #07111f;
    color: white;
    outline: none;
}

.input-group input:focus {
    border-color: #18d6a5;
}

.calculate {
    width: 100%;
    margin-top: 5px;
}

/* ================= MAP ================= */

.map-area {
    min-height: 530px;
    position: relative;
    overflow: hidden;
    border-radius: 14px;
    background:
        linear-gradient(#122438 1px, transparent 1px),
        linear-gradient(90deg, #122438 1px, transparent 1px);
    background-size: 40px 40px;
    border: 1px solid #1d3148;
}

.map-title {
    position: absolute;
    top: 15px;
    left: 18px;
    z-index: 10;
    background: rgba(7,17,31,0.9);
    padding: 8px 12px;
    border-radius: 7px;
    font-size: 13px;
}

.map-circle {
    position: absolute;
    border: 2px dashed;
    border-radius: 50%;
    transform: translate(-50%, -50%);
}

.satellite {
    position: absolute;
    transform: translate(-50%, -50%);
    z-index: 5;
    text-align: center;
}

.satellite-dot {
    width: 17px;
    height: 17px;
    border-radius: 50%;
    margin: auto;
    background: #3b82f6;
    box-shadow: 0 0 18px #3b82f6;
}

.satellite span {
    display: block;
    font-size: 12px;
    margin-top: 5px;
    color: #c5d5e7;
}

.target {
    position: absolute;
    width: 20px;
    height: 20px;
    border-radius: 50%;
    background: #18d6a5;
    box-shadow: 0 0 25px #18d6a5;
    transform: translate(-50%, -50%);
    z-index: 20;
}

.target::after {
    content: "";
    position: absolute;
    width: 36px;
    height: 36px;
    border: 2px solid #18d6a5;
    border-radius: 50%;
    left: -10px;
    top: -10px;
    animation: pulse 1.7s infinite;
}

@keyframes pulse {
    50% {
        transform: scale(1.3);
        opacity: 0.2;
    }
}

/* ================= RESULT ================= */

.result-grid {
    display: grid;
    grid-template-columns: repeat(3, 1fr);
    gap: 15px;
    margin-top: 20px;
}

.result-card {
    background: #07111f;
    border: 1px solid #1d3148;
    padding: 18px;
    border-radius: 10px;
}

.result-card small {
    color: #8ea4bd;
}

.result-card strong {
    display: block;
    font-size: 22px;
    color: #18d6a5;
}

/* ================= MATHEMATICS ================= */

.math-box {
    max-width: 900px;
    margin: auto;
    background: #0c1a2b;
    border: 1px solid #1d3148;
    padding: 30px;
    border-radius: 14px;
}

.formula {
    margin: 18px 0;
    padding: 18px;
    background: #07111f;
    border-left: 4px solid #18d6a5;
    border-radius: 5px;
    font-family: monospace;
    color: #d9e8f8;
    overflow-x: auto;
}

/* ================= SENSITIVITY ================= */

.sensitivity-grid {
    display: grid;
    grid-template-columns: 1fr 1fr;
    gap: 25px;
}

.range-box {
    background: #0c1a2b;
    border: 1px solid #1d3148;
    padding: 25px;
    border-radius: 14px;
}

.range-box input {
    width: 100%;
    accent-color: #18d6a5;
}

.range-value {
    font-size: 28px;
    color: #18d6a5;
    font-weight: bold;
    margin: 10px 0;
}

/* ================= CARDS ================= */

.cards {
    display: grid;
    grid-template-columns: repeat(3, 1fr);
    gap: 20px;
}

.info-card {
    background: #0c1a2b;
    border: 1px solid #1d3148;
    padding: 25px;
    border-radius: 14px;
    transition: 0.3s;
}

.info-card:hover {
    transform: translateY(-5px);
    border-color: #18d6a5;
}

.info-card .icon {
    font-size: 30px;
    margin-bottom: 15px;
}

.info-card h3 {
    margin-bottom: 10px;
}

.info-card p {
    color: #8ea4bd;
}

/* ================= FOOTER ================= */

footer {
    text-align: center;
    padding: 35px;
    border-top: 1px solid #1d3148;
    color: #71869d;
}

footer span {
    color: #18d6a5;
}

/* ================= RESPONSIVE ================= */

@media (max-width: 900px) {

    .team-info {
        display: none;
    }

    .menu-btn {
        display: block;
        margin-left: auto;
    }

    .nav {
        display: none;
        position: absolute;
        top: 72px;
        left: 0;
        right: 0;

        padding: 20px;

        flex-direction: column;
        background: #0c1a2b;
        border-bottom: 1px solid #1d3148;
    }

    .nav.show {
        display: flex;
    }

    .simulator-grid,
    .sensitivity-grid {
        grid-template-columns: 1fr;
    }

    .cards {
        grid-template-columns: 1fr;
    }
}

@media (max-width: 600px) {

    section {
        padding: 65px 0;
    }

    .header {
        padding: 0 20px;
    }

    .hero h1 {
        font-size: 45px;
    }

    .result-grid {
        grid-template-columns: 1fr;
    }

    .hero-buttons {
        flex-direction: column;
    }
}
</style>
</head>

<body>

<!-- ================= HEADER ================= -->

<header class="header">

    <div class="logo">
        <div class="logo-icon"></div>
        <span>Geo<span class="accent">Track</span></span>
    </div>

    <div class="team-info">
        <strong>TEAM A6</strong>
        <span>B.Tech Mathematics • GPS Trilateration</span>
    </div>

    <nav class="nav" id="nav">
        <a href="#simulator" class="active">Simulator</a>
        <a href="#analysis">Sensitivity</a>
        <a href="#mathematics">Mathematics</a>
        <a href="#how">How It Works</a>
        <a href="#applications">Applications</a>
    </nav>

    <button class="menu-btn" id="menuBtn"
            aria-label="Open navigation menu"
            aria-controls="nav"
            aria-expanded="false">
        ☰
    </button>

</header>


<!-- ================= HERO ================= -->

<section class="hero">

<div class="container">

    <div class="hero-content">

        <div class="badge">
            🛰️ GPS POSITIONING TECHNOLOGY
        </div>

        <h1>
            Locate Anything<br>
            With <span>Mathematics.</span>
        </h1>

        <p>
            GeoTrack demonstrates how GPS receivers calculate their
            position using trilateration, satellite distances and
            mathematical equations.
        </p>

        <div class="hero-buttons">

            <a href="#simulator"
               class="btn btn-primary">
                Launch Simulator
            </a>

            <a href="#mathematics"
               class="btn btn-secondary">
                Explore Mathematics
            </a>

        </div>

    </div>

</div>

</section>


<!-- ================= SIMULATOR ================= -->

<section id="simulator">

<div class="container">

<div class="section-title">

    <h2>GPS Trilateration Simulator</h2>

    <p>
        Adjust satellite coordinates and distances to estimate
        the receiver position.
    </p>

</div>


<div class="simulator-grid">

<!-- INPUT PANEL -->

<div class="panel">

<h3>Satellite Parameters</h3>

<div class="input-group">
<label>Satellite A — X</label>
<input id="x1" type="number" value="20">
</div>

<div class="input-group">
<label>Satellite A — Y</label>
<input id="y1" type="number" value="80">
</div>

<div class="input-group">
<label>Distance A</label>
<input id="r1" type="number" value="55">
</div>


<div class="input-group">
<label>Satellite B — X</label>
<input id="x2" type="number" value="80">
</div>

<div class="input-group">
<label>Satellite B — Y</label>
<input id="y2" type="number" value="75">
</div>

<div class="input-group">
<label>Distance B</label>
<input id="r2" type="number" value="45">
</div>


<div class="input-group">
<label>Satellite C — X</label>
<input id="x3" type="number" value="50">
</div>

<div class="input-group">
<label>Satellite C — Y</label>
<input id="y3" type="number" value="20">
</div>

<div class="input-group">
<label>Distance C</label>
<input id="r3" type="number" value="50">
</div>

<button class="btn btn-primary calculate"
        onclick="calculatePosition()">
    Calculate Position
</button>

</div>


<!-- MAP -->

<div class="map-area" id="map">

<div class="map-title">
    📡 Live Trilateration Map
</div>

<div id="circle1" class="map-circle"></div>
<div id="circle2" class="map-circle"></div>
<div id="circle3" class="map-circle"></div>

<div id="sat1" class="satellite">
    <div class="satellite-dot"></div>
    <span>GPS A</span>
</div>

<div id="sat2" class="satellite">
    <div class="satellite-dot"></div>
    <span>GPS B</span>
</div>

<div id="sat3" class="satellite">
    <div class="satellite-dot"></div>
    <span>GPS C</span>
</div>

<div id="target" class="target"></div>

</div>

</div>


<!-- RESULTS -->

<div class="result-grid">

<div class="result-card">
<small>Estimated X</small>
<strong id="resultX">--</strong>
</div>

<div class="result-card">
<small>Estimated Y</small>
<strong id="resultY">--</strong>
</div>

<div class="result-card">
<small>Position Error</small>
<strong id="error">--</strong>
</div>

</div>

</div>

</section>


<!-- ================= SENSITIVITY ================= -->

<section id="analysis">

<div class="container">

<div class="section-title">

<h2>Sensitivity Analysis</h2>

<p>
Understand how measurement errors affect GPS positioning.
</p>

</div>


<div class="sensitivity-grid">

<div class="range-box">

<h3>Distance Measurement Error</h3>

<div class="range-value">
<span id="errorValue">0</span>%
</div>

<input type="range"
       id="errorSlider"
       min="0"
       max="20"
       value="0">

<p>
GPS measurements are never perfectly accurate.
Even a small distance error can change the
calculated position.
</p>

</div>


<div class="range-box">

<h3>Estimated Accuracy</h3>

<div class="range-value">
<span id="accuracy">100</span>%
</div>

<p id="accuracyText">
The current measurements are assumed to be accurate.
</p>

</div>

</div>

</div>

</section>


<!-- ================= MATHEMATICS ================= -->

<section id="mathematics">

<div class="container">

<div class="section-title">

<h2>The Mathematics</h2>

<p>
GPS trilateration is based on distance equations.
</p>

</div>


<div class="math-box">

<h3>1. Circle Equation</h3>

<p>
Each satellite defines a circle around itself.
The receiver must lie somewhere on that circle.
</p>

<div class="formula">
(x - x₁)² + (y - y₁)² = r₁²
</div>


<h3>2. Three Satellite Equations</h3>

<div class="formula">
(x - x₁)² + (y - y₁)² = r₁²

<br><br>

(x - x₂)² + (y - y₂)² = r₂²

<br><br>

(x - x₃)² + (y - y₃)² = r₃²
</div>


<h3>3. Trilateration</h3>

<p>
The equations are solved together to find the
unknown receiver coordinates:
</p>

<div class="formula">
Receiver Position = (x, y)
</div>


<h3>4. GPS Concept</h3>

<p>
Real GPS systems use signals from multiple satellites.
The receiver estimates its distance from satellites
using signal travel time and then determines its position.
</p>

</div>

</div>

</section>


<!-- ================= HOW IT WORKS ================= -->

<section id="how">

<div class="container">

<div class="section-title">

<h2>How GeoTrack Works</h2>

<p>
The complete positioning process in four simple steps.
</p>

</div>


<div class="cards">

<div class="info-card">

<div class="icon">🛰️</div>

<h3>01. Satellite Signals</h3>

<p>
GPS satellites continuously transmit radio signals
containing timing and positioning information.
</p>

</div>


<div class="info-card">

<div class="icon">📏</div>

<h3>02. Distance Measurement</h3>

<p>
The receiver estimates its distance from each satellite
using the travel time of the signal.
</p>

</div>


<div class="info-card">

<div class="icon">📐</div>

<h3>03. Trilateration</h3>

<p>
The distances form intersecting circles.
Their intersection provides the estimated location.
</p>

</div>


<div class="info-card">

<div class="icon">📍</div>

<h3>04. Position</h3>

<p>
The calculated coordinates represent the receiver's
estimated geographic position.
</p>

</div>

</div>

</div>

</section>


<!-- ================= APPLICATIONS ================= -->

<section id="applications">

<div class="container">

<div class="section-title">

<h2>Real-World Applications</h2>

<p>
GPS trilateration is used in many modern technologies.
</p>

</div>


<div class="cards">

<div class="info-card">

<div class="icon">🚗</div>

<h3>Navigation</h3>

<p>
Google Maps, vehicle navigation and route planning
use satellite positioning.
</p>

</div>


<div class="info-card">

<div class="icon">📱</div>

<h3>Smartphones</h3>

<p>
Mobile devices use positioning technologies to
provide location-based services.
</p>

</div>


<div class="info-card">

<div class="icon">🚑</div>

<h3>Emergency Services</h3>

<p>
Location technology can help emergency responders
identify the approximate position of a caller.
</p>

</div>


<div class="info-card">

<div class="icon">🚢</div>

<h3>Transportation</h3>

<p>
Ships, aircraft and logistics systems use satellite
positioning for tracking and navigation.
</p>

</div>


<div class="info-card">

<div class="icon">🌾</div>

<h3>Precision Agriculture</h3>

<p>
GPS helps agricultural systems map fields and
guide machinery accurately.
</p>

</div>


<div class="info-card">

<div class="icon">🛰️</div>

<h3>Geospatial Systems</h3>

<p>
Surveying, mapping and geographic information systems
depend heavily on accurate positioning.
</p>

</div>

</div>

</div>

</section>


<!-- ================= FOOTER ================= -->

<footer>

Geo<span>Track</span> • TEAM A6

<br>

B.Tech Mathematics Project • GPS Trilateration

</footer>


<script>

/* ================= MOBILE MENU ================= */

const menuBtn = document.getElementById("menuBtn");
const nav = document.getElementById("nav");

menuBtn.addEventListener("click", function() {

    nav.classList.toggle("show");

    const opened = nav.classList.contains("show");

    menuBtn.setAttribute(
        "aria-expanded",
        opened
    );

});


/* CLOSE MENU AFTER CLICK */

document.querySelectorAll(".nav a").forEach(function(link) {

    link.addEventListener("click", function() {

        nav.classList.remove("show");

        menuBtn.setAttribute(
            "aria-expanded",
            "false"
        );

    });

});


/* ================= TRILATERATION ================= */

function calculatePosition() {

    const x1 = Number(document.getElementById("x1").value);
    const y1 = Number(document.getElementById("y1").value);
    const r1 = Number(document.getElementById("r1").value);

    const x2 = Number(document.getElementById("x2").value);
    const y2 = Number(document.getElementById("y2").value);
    const r2 = Number(document.getElementById("r2").value);

    const x3 = Number(document.getElementById("x3").value);
    const y3 = Number(document.getElementById("y3").value);
    const r3 = Number(document.getElementById("r3").value);


    /*
        Subtract equation 1 from equation 2:

        2(x2-x1)x + 2(y2-y1)y
        =
        r1²-r2²+x2²-x1²+y2²-y1²
    */

    const A = 2 * (x2 - x1);
    const B = 2 * (y2 - y1);

    const C =
        r1*r1 -
        r2*r2 +
        x2*x2 -
        x1*x1 +
        y2*y2 -
        y1*y1;


    /*
        Subtract equation 1 from equation 3.
    */

    const D = 2 * (x3 - x1);
    const E = 2 * (y3 - y1);

    const F =
        r1*r1 -
        r3*r3 +
        x3*x3 -
        x1*x1 +
        y3*y3 -
        y1*y1;


    const denominator = A * E - B * D;


    if (Math.abs(denominator) < 0.0001) {

        alert(
            "The satellite geometry is not suitable for trilateration."
        );

        return;
    }


    const x = (C * E - B * F) / denominator;

    const y = (A * F - C * D) / denominator;


    document.getElementById("resultX").textContent =
        x.toFixed(2);

    document.getElementById("resultY").textContent =
        y.toFixed(2);


    /* Calculate consistency error */

    const d1 = Math.sqrt(
        Math.pow(x-x1,2) +
        Math.pow(y-y1,2)
    );

    const d2 = Math.sqrt(
        Math.pow(x-x2,2) +
        Math.pow(y-y2,2)
    );

    const d3 = Math.sqrt(
        Math.pow(x-x3,2) +
        Math.pow(y-y3,2)
    );


    const error =
        (
            Math.abs(d1-r1) +
            Math.abs(d2-r2) +
            Math.abs(d3-r3)
        ) / 3;


    document.getElementById("error").textContent =
        error.toFixed(2);


    /* Draw satellites and circles */

    drawMap(
        x1,y1,r1,
        x2,y2,r2,
        x3,y3,r3,
        x,y
    );
}


/* ================= MAP DRAWING ================= */

function drawMap(
    x1,y1,r1,
    x2,y2,r2,
    x3,y3,r3,
    x,y
) {

    const map = document.getElementById("map");

    const width = map.clientWidth;
    const height = map.clientHeight;


    function pxX(value) {

        return (
            value / 100
        ) * width;

    }


    function pxY(value) {

        return (
            1 - value / 100
        ) * height;

    }


    function setSatellite(
        id,
        sx,
        sy
    ) {

        const element =
            document.getElementById(id);

        element.style.left =
            pxX(sx) + "px";

        element.style.top =
            pxY(sy) + "px";

    }


    function setCircle(
        id,
        sx,
        sy,
        radius
    ) {

        const circle =
            document.getElementById(id);

        const diameter =
            (radius / 100) *
            Math.min(width,height) * 2;

        circle.style.width =
            diameter + "px";

        circle.style.height =
            diameter + "px";

        circle.style.left =
            pxX(sx) + "px";

        circle.style.top =
            pxY(sy) + "px";

    }


    setSatellite("sat1",x1,y1);
    setSatellite("sat2",x2,y2);
    setSatellite("sat3",x3,y3);


    setCircle("circle1",x1,y1,r1);
    setCircle("circle2",x2,y2,r2);
    setCircle("circle3",x3,y3,r3);


    const target =
        document.getElementById("target");

    target.style.left =
        pxX(x) + "px";

    target.style.top =
        pxY(y) + "px";

}


/* ================= SENSITIVITY ================= */

const errorSlider =
    document.getElementById("errorSlider");

const errorValue =
    document.getElementById("errorValue");

const accuracy =
    document.getElementById("accuracy");

const accuracyText =
    document.getElementById("accuracyText");


errorSlider.addEventListener(
    "input",
    function() {

        const value =
            Number(this.value);

        errorValue.textContent =
            value;

        const calculatedAccuracy =
            Math.max(
                0,
                100 - value * 3
            );

        accuracy.textContent =
            calculatedAccuracy.toFixed(0);


        if (value === 0) {

            accuracyText.textContent =
                "The current measurements are assumed to be accurate.";

        }

        else if (value <= 5) {

            accuracyText.textContent =
                "Small measurement errors may produce small position errors.";

        }

        else if (value <= 10) {

            accuracyText.textContent =
                "Moderate measurement errors can noticeably affect the position.";

        }

        else {

            accuracyText.textContent =
                "Large measurement errors can significantly reduce positioning accuracy.";

        }

    }
);


/* ================= INITIAL SIMULATION ================= */

window.addEventListener(
    "load",
    calculatePosition
);


/* ================= RESIZE ================= */

window.addEventListener(
    "resize",
    calculatePosition
);

</script>

</body>
</html>
'''

# Save website
file_path = "/content/geotrack.html"

with open(file_path, "w", encoding="utf-8") as f:
    f.write(html)

# Display website inside Colab
display(
    IFrame(
        src=file_path,
        width="100%",
        height=900,
        scrolling=True
    )
)

print("✅ GeoTrack website launched successfully!")
print("📍 GPS Trilateration Simulator is ready.")
```

```python
%%writefile /content/script.js

/* =========================================
   GEOTRACK GPS TRILATERATION
   TEAM A6
========================================= */


/* =========================================
   MOBILE NAVIGATION
========================================= */

const menuBtn =
    document.getElementById("menuBtn");

const nav =
    document.getElementById("nav");


menuBtn.addEventListener(
    "click",
    function () {

        nav.classList.toggle("show");

        const opened =
            nav.classList.contains("show");

        menuBtn.setAttribute(
            "aria-expanded",
            opened
        );

    }
);


/* Close menu after clicking a link */

document
    .querySelectorAll(".nav a")
    .forEach(function (link) {

        link.addEventListener(
            "click",
            function () {

                nav.classList.remove("show");

                menuBtn.setAttribute(
                    "aria-expanded",
                    "false"
                );

            }
        );

    });



/* =========================================
   GET INPUT VALUE
========================================= */

function getValue(id) {

    return Number(
        document.getElementById(id).value
    );

}



/* =========================================
   MAIN TRILATERATION
========================================= */

function calculatePosition() {


    /* Satellite A */

    const x1 = getValue("x1");
    const y1 = getValue("y1");
    const r1 = getValue("r1");


    /* Satellite B */

    const x2 = getValue("x2");
    const y2 = getValue("y2");
    const r2 = getValue("r2");


    /* Satellite C */

    const x3 = getValue("x3");
    const y3 = getValue("y3");
    const r3 = getValue("r3");



    /*
       Circle equations:

       (x-x1)^2 + (y-y1)^2 = r1^2

       (x-x2)^2 + (y-y2)^2 = r2^2

       (x-x3)^2 + (y-y3)^2 = r3^2
    */


    /*
       Subtract equation 1 from equation 2.
    */

    const A =
        2 * (x2 - x1);

    const B =
        2 * (y2 - y1);

    const C =
        r1 * r1
        - r2 * r2
        + x2 * x2
        - x1 * x1
        + y2 * y2
        - y1 * y1;



    /*
       Subtract equation 1 from equation 3.
    */

    const D =
        2 * (x3 - x1);

    const E =
        2 * (y3 - y1);

    const F =
        r1 * r1
        - r3 * r3
        + x3 * x3
        - x1 * x1
        + y3 * y3
        - y1 * y1;



    /*
       Calculate determinant.
    */

    const denominator =
        A * E - B * D;



    /*
       Check satellite geometry.
    */

    if (
        Math.abs(denominator)
        < 0.000001
    ) {

        alert(
            "The satellite geometry is not suitable for trilateration."
        );

        return;

    }



    /*
       Solve for X and Y.
    */

    const x =
        (C * E - B * F)
        / denominator;


    const y =
        (A * F - C * D)
        / denominator;



    /*
       Display calculated position.
    */

    document
        .getElementById("resultX")
        .textContent =
        x.toFixed(2);


    document
        .getElementById("resultY")
        .textContent =
        y.toFixed(2);



    /*
       Calculate distance errors.
    */

    const distance1 =
        Math.sqrt(
            Math.pow(x - x1, 2)
            +
            Math.pow(y - y1, 2)
        );


    const distance2 =
        Math.sqrt(
            Math.pow(x - x2, 2)
            +
            Math.pow(y - y2, 2)
        );


    const distance3 =
        Math.sqrt(
            Math.pow(x - x3, 2)
            +
            Math.pow(y - y3, 2)
        );



    /*
       Average distance error.
    */

    const averageError =

        (
            Math.abs(distance1 - r1)
            +
            Math.abs(distance2 - r2)
            +
            Math.abs(distance3 - r3)
        ) / 3;



    document
        .getElementById("error")
        .textContent =
        averageError.toFixed(2);



    /*
       Draw the map.
    */

    drawMap(
        x1,
        y1,
        r1,

        x2,
        y2,
        r2,

        x3,
        y3,
        r3,

        x,
        y
    );

}



/* =========================================
   DRAW MAP
========================================= */

function drawMap(

    x1, y1, r1,

    x2, y2, r2,

    x3, y3, r3,

    x, y

) {


    const map =
        document.getElementById("map");


    const width =
        map.clientWidth;


    const height =
        map.clientHeight;



    /*
       Convert mathematical X coordinate
       into screen X coordinate.
    */

    function convertX(value) {

        return (
            value / 100
        ) * width;

    }



    /*
       Convert mathematical Y coordinate
       into screen Y coordinate.

       Browser Y increases downward,
       therefore we reverse it.
    */

    function convertY(value) {

        return (
            1 - value / 100
        ) * height;

    }



    /*
       Position satellite.
    */

    function positionSatellite(
        id,
        sx,
        sy
    ) {

        const element =
            document.getElementById(id);


        element.style.left =
            convertX(sx) + "px";


        element.style.top =
            convertY(sy) + "px";

    }



    /*
       Position circle.
    */

    function positionCircle(
        id,
        sx,
        sy,
        radius
    ) {

        const circle =
            document.getElementById(id);


        /*
           Convert mathematical radius
           into screen diameter.
        */

        const diameter =

            (
                radius / 100
            )
            *
            Math.min(
                width,
                height
            )
            *
            2;


        circle.style.width =
            diameter + "px";


        circle.style.height =
            diameter + "px";


        circle.style.left =
            convertX(sx) + "px";


        circle.style.top =
            convertY(sy) + "px";

    }



    /* Satellites */

    positionSatellite(
        "sat1",
        x1,
        y1
    );


    positionSatellite(
        "sat2",
        x2,
        y2
    );


    positionSatellite(
        "sat3",
        x3,
        y3
    );



    /* Circles */

    positionCircle(
        "circle1",
        x1,
        y1,
        r1
    );


    positionCircle(
        "circle2",
        x2,
        y2,
        r2
    );


    positionCircle(
        "circle3",
        x3,
        y3,
        r3
    );



    /* Receiver */

    const target =
        document.getElementById(
            "target"
        );


    target.style.left =
        convertX(x) + "px";


    target.style.top =
        convertY(y) + "px";

}



/* =========================================
   SENSITIVITY ANALYSIS
========================================= */

const errorSlider =
    document.getElementById(
        "errorSlider"
    );


const errorValue =
    document.getElementById(
        "errorValue"
    );


const accuracy =
    document.getElementById(
        "accuracy"
    );


const accuracyText =
    document.getElementById(
        "accuracyText"
    );



errorSlider.addEventListener(
    "input",
    function () {


        const value =
            Number(this.value);


        /*
           Display measurement error.
        */

        errorValue.textContent =
            value;



        /*
           Simple educational
           accuracy model.
        */

        const calculatedAccuracy =

            Math.max(
                0,
                100 - value * 3
            );


        accuracy.textContent =
            calculatedAccuracy
                .toFixed(0);



        /*
           Display explanation.
        */

        if (value === 0) {

            accuracyText.textContent =
                "The current measurements are assumed to be accurate.";

        }

        else if (value <= 5) {

            accuracyText.textContent =
                "Small measurement errors may produce small position errors.";

        }

        else if (value <= 10) {

            accuracyText.textContent =
                "Moderate measurement errors can noticeably affect the position.";

        }

        else {

            accuracyText.textContent =
                "Large measurement errors can significantly reduce positioning accuracy.";

        }

    }
);



/* =========================================
   CALCULATE BUTTON
========================================= */

document
    .getElementById("calculateBtn")
    .addEventListener(
        "click",
        calculatePosition
    );



/* =========================================
   INITIAL CALCULATION
========================================= */

window.addEventListener(
    "load",
    function () {

        calculatePosition();

    }
);



/* =========================================
   REDRAW MAP WHEN WINDOW RESIZES
========================================= */

window.addEventListener(
    "resize",
    function () {

        calculatePosition();

    }
);
```

```python
# ============================================
# GOOGLE COLAB - RUN HTML/CSS/JS WEBSITE
# ============================================

import os
import socket
import threading
from http.server import ThreadingHTTPServer, SimpleHTTPRequestHandler
from google.colab.output import eval_js
from IPython.display import display, HTML

# ---------- Find index.html ----------
SEARCH_DIRS = ["/content", "/content/drive/MyDrive"]

index_file = None

for base in SEARCH_DIRS:
    if os.path.exists(base):
        for root, dirs, files in os.walk(base):
            if "index.html" in files:
                index_file = os.path.join(root, "index.html")
                break
    if index_file:
        break

if index_file is None:
    raise FileNotFoundError(
        "index.html was not found. Please upload your HTML/CSS/JS files to Colab first."
    )

# Website folder
WEB_DIR = os.path.dirname(index_file)

print("Website found:")
print(WEB_DIR)

# ---------- Choose an available port ----------
def find_free_port(start=8000):
    for port in range(start, start + 100):
        with socket.socket(socket.AF_INET, socket.SOCK_STREAM) as s:
            try:
                s.bind(("127.0.0.1", port))
                return port
            except OSError:
                continue
    raise RuntimeError("No free port found.")

PORT = find_free_port()

# ---------- Change to website directory ----------
os.chdir(WEB_DIR)

# ---------- Web server ----------
class WebsiteHandler(SimpleHTTPRequestHandler):
    def log_message(self, format, *args):
        pass  # Keep Colab output clean

server = ThreadingHTTPServer(
    ("0.0.0.0", PORT),
    WebsiteHandler
)

# Run server in background
thread = threading.Thread(
    target=server.serve_forever,
    daemon=True
)
thread.start()

# ---------- Create Colab browser URL ----------
url = eval_js(f"google.colab.kernel.proxyPort({PORT})")

print("\n============================================")
print("WEBSITE SERVER STARTED")
print("============================================")
print("Port:", PORT)
print("Website:", url)
print("============================================\n")

# ---------- Clickable button ----------
display(HTML(f"""
<div style="
    padding:20px;
    background:#111827;
    border-radius:15px;
    margin:10px 0;
    font-family:Arial;
">
    <h2 style="color:white;margin-top:0;">
        🚀 Your Website is Running
    </h2>

    <a href="{url}" target="_blank"
       style="
       display:inline-block;
       padding:12px 22px;
       background:#2563eb;
       color:white;
       text-decoration:none;
       border-radius:10px;
       font-weight:bold;
       ">
       OPEN WEBSITE
    </a>

    <p style="color:#9ca3af;margin-bottom:0;">
        Click the button above to open your HTML/CSS/JS website.
    </p>
</div>
"""))

# ---------- Display website inside Colab ----------
display(HTML(f"""
<iframe
    src="{url}"
    width="100%"
    height="800"
    style="
        border:0;
        border-radius:15px;
        background:white;
    ">
</iframe>
"""))
```
